```python
# Assignment 1: Calculate Undiscounted and Discounted Returns


# Step 1: Accept number of time steps
n = int(input("Enter the number of time steps: "))

# Step 2: Accept rewards
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward at time step {i}: "))
    rewards.append(reward)

# Step 3: Calculate undiscounted return
undiscounted_return = sum(rewards)

# Step 4: Accept discount factor
gamma = float(input("Enter the discount factor (gamma): "))

# Check discount factor
if gamma < 0 or gamma > 1:
    print("Discount factor must be between 0 and 1.")
else:
    # Step 5: Calculate discounted return
    discounted_return = 0

    for i in range(n):
        discounted_return += rewards[i] * (gamma ** i)

    # Step 6: Display results
    print("\n----- RESULTS -----")
    print("Entered reward values:", rewards)
    print("Undiscounted Return:", undiscounted_return)
    print("Discounted Return:", round(discounted_return, 2))
```

```python
# Assignment 2: 2 x 3 Grid World as an MDP



# --------------------------------------------------
# 1. Define State Space
# --------------------------------------------------

states = ["S0", "S1", "G", "X", "S3", "S4"]

print("\nState Space:")
print(states)

# --------------------------------------------------
# 2. Define Action Space
# --------------------------------------------------

actions = ["Up", "Down", "Left", "Right"]

print("\nAction Space:")
print(actions)

# --------------------------------------------------
# 3. Define Reward Function
# --------------------------------------------------

rewards = {
    "S0": -1,
    "S1": -1,
    "S3": -1,
    "S4": -1,
    "G": 10,
    "X": -5
}

print("\nReward Function:")
for state in states:
    print(state, "=", rewards[state])

# --------------------------------------------------
# 4. Define Discount Factor
# --------------------------------------------------

gamma = 0.9

print("\nDiscount Factor (gamma):", gamma)

# --------------------------------------------------
# 5. Define Transition Function
# --------------------------------------------------

# Transition table based on the given 2 x 3 Grid World

transition = {
    "S0": {
        "Up": "S0",
        "Down": "S0",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "X",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    },

    "G": {
        "Up": "G",
        "Down": "G",
        "Left": "G",
        "Right": "G"
    },

    "X": {
        "Up": "X",
        "Down": "X",
        "Left": "X",
        "Right": "X"
    }
}

# --------------------------------------------------
# Display Transition Function
# --------------------------------------------------

print("\nTransition Function:")

for state in states:
    print("\nState:", state)

    for action in actions:
        print("  ", action, "->", transition[state][action])

# --------------------------------------------------
# 6. Estimate Next State
# --------------------------------------------------

print("\n--------------------")
print("NEXT STATE PREDICTION")
print("--------------------")

current_state = input(
    "Enter current state (S0, S1, S3, S4, G, X): "
)

action = input(
    "Enter action (Up, Down, Left, Right): "
)

# Validate state and action

if current_state not in states:
    print("Invalid state!")

elif action not in actions:
    print("Invalid action!")

else:
    next_state = transition[current_state][action]

    print("\nCurrent State :", current_state)
    print("Action        :", action)
    print("Next State    :", next_state)

    # Display reward of next state
    print("Reward        :", rewards[next_state])

    # Check terminal state
    if next_state == "G":
        print("Result        : Goal reached! Episode terminates.")

    elif next_state == "X":
        print("Result        : Obstacle reached! Episode terminates.")

    else:
        print("Result        : Continue to next state.")
```
